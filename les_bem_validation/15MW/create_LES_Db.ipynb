## Create LES inputs for 15MW turbine

```python
%load_ext autoreload
%autoreload 2
```

```python
import numpy as np
from MITRotor import IEA15MW
```

```python
rotor = IEA15MW()
rotor.R*2
```

```python
rotor.hub_radius*2
```

```python
# Airfoil names
airfoil_names = np.array(['circular', 'circular', 'SNL-FFA-W3-500', 'FFA-W3-360', 'FFA-W3-330blend', 'FFA-W3-301', 'FFA-W3-270blend', 'FFA-W3-241', 'FFA-W3-211', 'FFA-W3-211'])
airfoil_names
```

```python
# Airfoil indices cooresponding to demarkations
airfoil_index = np.array([1,1,2,3,4,5,6,7,8,8])
airfoil_index
```

```python
len(airfoil_names) == len(airfoil_index)
```

```python
# Nondimensional airfoil demarkations
airfoil_mu = np.array([0.03264463, 0.05199174, 0.17774793, 0.26981145, 0.35075359, 0.45748713, 0.55276395, 0.65001823, 0.77919519, 1.0])
airfoil_mu
```

```python
# Dimensional airfoil demarkations
airfoil_r = airfoil_mu * rotor.R
airfoil_r
```

```python
# MITRotor twist
twist = np.rad2deg(rotor.twist_func(airfoil_mu))
twist
```

```python
# MITRotor thickness
thickness = rotor.solidity_func(airfoil_mu)
thickness
```

```python
# MITRotor chord length
chord = rotor.chord_func(airfoil_mu)
chord
```

```python
# Turbine database
for i in range(len(airfoil_r)):
    print("{:7.3f} {:7.4f} {:7.4f} {:8.4f}  {:<18} {:1.0f}".format(
    airfoil_r[i], chord[i], thickness[i], twist[i], f'"{airfoil_names[i]}"', airfoil_index[i]
))
```

```python
aoa = np.linspace(-np.pi, np.pi, 120)
```

```python
string = 'SNL-FFA-W3-500'

index = 2

ref_CL = rotor.airfoil_func.cl_2d_raw[index,:]
ref_CD = rotor.airfoil_func.cd_2d_raw[index,:]

with open(f'./airfoil_tables/{string}.dat', "w") as f:
    f.write(f'{len(aoa)} "Cl" "Cd" "Cm" GENERATED FROM MITROTOR\n')
    for i in range(len(aoa)):
        f.write("{:8.3f}  {:7.4f}  {:7.4f}  {:7.4f}\n".format(
        np.rad2deg(aoa[i]), ref_CL[i], ref_CD[i], 0.0
    ))
```
