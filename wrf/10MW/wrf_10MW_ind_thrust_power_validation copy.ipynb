```python
casenames = [r'U06', r'U07', r'U08', r'U09', r'U10']
# casenames = [r's0_v4']

wrfles_data = []
# wrfles_wind = []
for count, name in enumerate(casenames):
    wrfles_data.append(dict(np.load('/scratch/09909/smata/wrf_les_sweep/runs/shapiro_testing/'+casenames[count]+'_lite.npz')))
```

```python
rotor = IEA10MW()

nElements = 26
nSections = 158
```

```python
def extract_columns(file_list):
    data_dict = {}
    
    for file in file_list:
        with open(file, 'r') as f:
            lines = f.readlines()
            
            # Skip the header line and extract columns 1, 4, and 5
            data = np.loadtxt(lines[1:])[:, [0, 3, 4]]
            
            # Use the filename without the path and extension as the dictionary key
            key = Path(file).parent.name
            data_dict[key] = data
    
    return data_dict
```

```python
file_list = glob.glob("/scratch/09909/smata/wrf_les_sweep/runs/shapiro_testing/*/input_sounding")

data = extract_columns(file_list)

t = np.linspace(0,2*np.pi,nSections)
r = np.linspace(0,1,nElements)

R, T = np.meshgrid(r, t)

X = R * np.sin(T)
Y = (R * np.cos(T)) * wrfles_data[0]['diameter']/2 + 378
```

```python
offlinePower  = np.zeros_like(casenames, dtype='float')
thrust        = np.zeros_like(casenames, dtype='float')
power         = np.zeros_like(casenames, dtype='float')
IntThrust_old = np.zeros_like(casenames, dtype='float')
IntThrust_new = np.zeros_like(casenames, dtype='float')

offlinethrustsum = np.zeros_like(casenames, dtype='float')

offlineThrust_old= np.zeros_like(casenames, dtype='float')
offlineThrust_new= np.zeros_like(casenames, dtype='float')

chord = [4.649046,4.844576,5.216862,5.647547,5.931514,5.96353,5.812206,5.569981,5.273268,4.94348,4.598646,4.251929,3.911027,3.581805,3.269938,2.978544,2.70897,2.462014,2.237137,2.035348,1.855479,1.691271,1.526768,1.308778,0.7955106,0.5734]

chord =  (np.ones_like(Cd.T) * chord).T
chord_new =  (np.ones_like(Cd.T) * chord).T

for i in range(len(casenames)):

    u_func = interp1d(data[casenames[i]][:,0], data[casenames[i]][:,1], kind='linear')
    v_func = interp1d(data[casenames[i]][:,0], data[casenames[i]][:,2], kind='linear')

    u_inf = u_func(Y)
    v_inf = v_func(Y)

    # Get radial points
    a = ((wrfles_data[i]['rOverR'] * rotor.R) - rotor.hub_radius)/rotor.R
    r = (rotor.hub_radius + a * (rotor.R - rotor.hub_radius)) / rotor.R

    phi = np.deg2rad(np.mean(wrfles_data[i]['phi'], axis=0))
    Cl = np.mean(wrfles_data[i]['cl'], axis=0)
    Cd = np.mean(wrfles_data[i]['cd'], axis=0)

    r =  (np.ones_like(Cd.T) * r).T

    chord_old = rotor.chord_func(r)

    # Axial coefficient
    Cax = Cl * np.cos(phi) + Cd * np.sin(phi)

    L = 0.5 * 1.225 * Cl * chord * np.mean(wrfles_data[i]['vrel'], axis=0)**2
    D = 0.5 * 1.225 * Cd * chord * np.mean(wrfles_data[i]['vrel'], axis=0)**2

    Fn = L * np.cos(phi) + D * np.sin(phi)
    Ft = L * np.sin(phi) - D * np.cos(phi)

    Ctan = Cl * np.sin(phi) - Cd * np.cos(phi)

    dT = 1/2 * 3 * 1.225 * chord * np.mean(wrfles_data[i]['vrel'], axis=0)**2 * Cax * dr
    dT_old = 1/2 * 3 * 1.225 * chord_old * np.mean(wrfles_data[i]['vrel'], axis=0)**2 * Cax 
    dT_new = 1/2 * 3 * 1.225 * chord_new * np.mean(wrfles_data[i]['vrel'], axis=0)**2 * Cax 

    dr = ( rotor.R - rotor.hub_radius)/nElements

    dN = 3 * Fn * dr

    # print(f'vrel: {np.mean(wrfles_data[i]["vrel"], axis=0)}')
    # print(f'Cax: {Cax}')
    # print(f'chord: {chord}')
    # print(f'cl: {Cl}')
    # print(f'cd: {Cd}')

    dQ = 1/2 * 3 * 1.225 * chord * np.mean(wrfles_data[i]['vrel'], axis=0)**2 * Ctan * r

    IntThrust_old[i] = 1/(2*np.pi) * np.trapezoid(np.trapezoid(dT_old,np.linspace(0, 2*np.pi, nSections)),np.linspace(0,1,26)*199/2)/1000
    IntThrust_new[i] = 1/(2*np.pi) * np.trapezoid(np.trapezoid(dT_new,np.linspace(0, 2*np.pi, nSections)),np.linspace(0,1,26)*199/2)/1000

    # IntThrust[i] = 1/(2*np.pi) * np.trapezoid(np.trapezoid(dT,np.linspace(0, 2*np.pi, nSections)),np.linspace(0,1,26)*199/2)/1000


    offlineThrust_old[i] = np.sum(dT_old*dr) / 1000 / nSections
    offlineThrust_new[i] = np.sum(dT_new*dr) / 1000 / nSections

    offlinethrustsum[i] = np.sum((dN)) / nSections / 1000

    offlinePower[i] = 1/(2*np.pi) * np.trapezoid(np.trapezoid(dQ,np.linspace(0, 2*np.pi, nSections)),np.linspace(0,1,26)) * np.mean(wrfles_data[i]['omega'])*0.944

    thrust[i]    = np.mean(wrfles_data[i]['thrust'],axis=0).item()/1000
    power[i]     = np.mean(wrfles_data[i]['power_mech'],axis=0).item()/1000
```
