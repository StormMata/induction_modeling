```python
import numpy as np
from scipy.io import loadmat, savemat
```

```python
save_path = '/scratch/09909/smata/induction_modeling/gaussian_process/all_turbines/train_data/'
```

```python
path_10MW = '/scratch/09909/smata/induction_modeling/gaussian_process/10MW/train_data/'
path_15MW = '/scratch/09909/smata/induction_modeling/gaussian_process/15MW/train_data/'
path_22MW = '/scratch/09909/smata/induction_modeling/gaussian_process/22MW/train_data/'
```

# RAW data

```python
# Annulus
mat_10MW_ann = loadmat(path_10MW + 'wrf_10MW_ann_RAW.mat')
mat_15MW_ann = loadmat(path_15MW + 'wrf_15MW_ann_RAW.mat')
mat_22MW_ann = loadmat(path_22MW + 'wrf_22MW_ann_RAW.mat')


X_10MW_ann = mat_10MW_ann['X']
X_15MW_ann = mat_15MW_ann['X']
X_22MW_ann = mat_22MW_ann['X']

y_10MW_ann = mat_10MW_ann['y']
y_15MW_ann = mat_15MW_ann['y']
y_22MW_ann = mat_22MW_ann['y']

# Stack vertically
X_all = np.vstack([X_10MW_ann, X_15MW_ann, X_22MW_ann])
y_all = np.vstack([y_10MW_ann, y_15MW_ann, y_22MW_ann])

# Save the combined data
savemat(save_path + 'wrf_all_ann_RAW.mat', {'X': X_all, 'y': y_all})
```

```python
# Rotor
mat_10MW_rot = loadmat(path_10MW + 'wrf_10MW_rot_RAW.mat')
mat_15MW_rot = loadmat(path_15MW + 'wrf_15MW_rot_RAW.mat')
mat_22MW_rot = loadmat(path_22MW + 'wrf_22MW_rot_RAW.mat')

X_10MW_rot = mat_10MW_rot['X']
X_15MW_rot = mat_15MW_rot['X']
X_22MW_rot = mat_22MW_rot['X']

y_10MW_rot = mat_10MW_rot['y']
y_15MW_rot = mat_15MW_rot['y']
y_22MW_rot = mat_22MW_rot['y']

# Stack vertically
X_all = np.vstack([X_10MW_rot, X_15MW_rot, X_22MW_rot])
y_all = np.vstack([y_10MW_rot, y_15MW_rot, y_22MW_rot])

# Save the combined data
savemat(save_path + 'wrf_all_rot_RAW.mat', {'X': X_all, 'y': y_all})
```
